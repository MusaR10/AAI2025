```python
  import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
# Generate sample customer data
#data = {
#'avg_order_value': [500, 1200, 300, 1500, 800, 200, 1000, 600, 1300, 400],
#'purchase_frequency': [5, 12, 3, 15, 8, 2, 10, 6, 13, 4],
#'spending_score	': [25, 34, 45, 28, 52, 36, 41, 29, 47, 33],
#'region': ['North', 'South', 'West', 'East', 'South', 'North', 'West', 'East',
#'South', 'North']
#}
df = pd.read_csv('/content/ecommerce_customers.csv')
# data from https://github.com/rskworld/ecommerce-customers/blob/main/ecommerce_customers.csv?
# Preprocess data: Select numerical features and scale them
features = ['avg_order_value', 'purchase_frequency', 'spending_score']
X = df[features]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Determine optimal number of clusters using elbow method
inertia = []
K = range(1, 6)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
# Plot elbow curve
plt.figure(figsize=(8, 5))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.savefig('elbow_plot.png')
plt.close()
# Apply K-Means with optimal K (e.g., 3 based on elbow method)
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
df['cluster'] = kmeans.fit_predict(X_scaled)
# Analyze clusters
cluster_summary = df.groupby('cluster')[features].mean().round(2)
print("Cluster Characteristics:")
print(cluster_summary)
# Example of targeted strategies
for cluster in range(optimal_k):
    print(f"\nCluster {cluster} Strategy:")
    if cluster_summary.loc[cluster, 'avg_order_value'] > 170:
        print("High-spending customers: Offer exclusive promotions or loyalty rewards. They are spending a lot each time they purchse, however they arent purchasing frequently. Try to increase the frequency of purchase")
    elif cluster_summary.loc[cluster, 'purchase_frequency'] > 12:
        print("Frequent buyers: Provide bulk discounts in order to try and get larger orders from them. Try to increase the average value")
    else:
        print("Low-engagement customers: Send personalized re-engagement campaigns, they arent spending the most or have the most frequency. Try to partially improve both")
# Save cluster assignments to CSV
df.to_csv('customer_segments.csv', index=False)
```
